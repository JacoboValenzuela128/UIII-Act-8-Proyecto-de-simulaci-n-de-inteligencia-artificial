# VETERINARIO Y ESTETOSCOPIO

```python
# ==========================================================
# SIMULADOR IA: VET-SCOPE SMART (Veterinario & Estetoscopio)
# ==========================================================

print("==================================================")
print("     ¡BIENVENIDO A VET-SCOPE SMART v2.0!          ")
print("  El sistema de IA para Estetoscopios Veterinarios ")
print("==================================================")
print("Coloca el estetoscopio digital en el paciente...")

# Iniciamos el ciclo infinito interactivo
while True:
    # Impresión del Menú Principal
    print("\n--- MENÚ DEL VETERINARIO SMART ---")
    print("1. Analizar ritmo cardíaco (Latidos por minuto)")
    print("2. Analizar sonido respiratorio (Pulmones)")
    print("3. Apagar el Estetoscopio Inteligente (Salir)")

    opcion = input("\nSelecciona una opción del menú (1-3): ")

    # ------------------------------------------------------
    # OPCIÓN 1: ANALIZAR RITMO CARDÍACO
    # ------------------------------------------------------
    if opcion == "1":
        print("\n--- MÓDULO DE CARDIOLOGÍA ANIMAL ---")
        tipo_mascota = input("¿El paciente es un (perro) o un (gato)?: ")
        latidos = input("¿Cuántos latidos por minuto (LPM) registró el estetoscopio?: ")
        latidos = int(latidos) # Convertimos el texto a número entero

        print("\n[IA procesando frecuencias cardíacas...]")

        # Lógica de IA para PERROS
        if tipo_mascota == "perro":
            if latidos < 60:
                print("🚨 ALERTA IA: Bradicardia (Ritmo cardíaco demasiado bajo).")
                print("🩺 Sugerencia Veterinaria: Revisar posible hipotermia o problemas del corazón.")
            elif latidos >= 60 and latidos <= 140:
                print("✅ IA Info: Ritmo cardíaco NORMAL para un perro.")
                print("🩺 Sugerencia Veterinaria: El corazón de la mascota se escucha fuerte y sano.")
            else:
                print("🚨 ALERTA IA: Taquicardia (Ritmo cardíaco muy elevado).")
                print("🩺 Sugerencia Veterinaria: El paciente puede tener estrés, dolor o fiebre.")

        # Lógica de IA para GATOS
        elif tipo_mascota == "gato":
            if latidos < 140:
                print("🚨 ALERTA IA: Ritmo cardíaco peligrosamente bajo para un felino.")
            elif latidos >= 140 and latidos <= 220:
                print("✅ IA Info: Ritmo cardíaco NORMAL para un gato (los gatos tienen el corazón más rápido).")
            else:
                print("🚨 ALERTA IA: Taquicardia felina. ¡Nivel de estrés o soplo cardíaco alto!")

        # Si no escriben perro o gato
        else:
            print("❌ Especie no registrada en la base de datos de esta IA.")

    # ------------------------------------------------------
    # OPCIÓN 2: ANALIZAR SONIDO RESPIRATORIO
    # ------------------------------------------------------
    elif opcion == "2":
        print("\n--- MÓDULO RESPIRATORIO (Auscultación) ---")
        print("Escucha con atención el estetoscopio e identifica el sonido dominante:")
        print("- (normal): Sonido limpio de aire.")
        print("- (silbido): Sonido agudo al respirar.")
        print("- (burbujeo): Sonido húmedo o crujiente.")

        sonido = input("¿Qué sonido detectas?: ")

        print("\n[IA Analizando frecuencias acústicas...]")

        if sonido == "normal":
            print("✅ IA Info: Pulmones limpios. El flujo de aire es perfecto.")
        elif sonido == "silbido":
            print("⚠️ DIAGNÓSTICO IA: Posible obstrucción de vías respiratorias o asma felina/canina.")
            print("🩺 Sugerencia: Se recomienda tratamiento con nebulizaciones.")
        elif sonido == "burbujeo":
            print("🚨 DIAGNÓSTICO CRÍTICO IA: Presencia de líquido en los pulmones (Posible neumonía).")
            print("🩺 Sugerencia: Requiere atención médica inmediata y radiografía de tórax.")
        else:
            print("❌ Sonido no reconocido por los filtros del estetoscopio.")

    # ------------------------------------------------------
    # OPCIÓN 3: APAGAR EL DISPOSITIVO (SALIR)
    # ------------------------------------------------------
    elif opcion == "3":
        print("\nApagando Vet-Scope Smart... Guardando datos en la nube médica.")
        break  # <---- ¡Aquí rompemos el ciclo infinito de manera limpia!

    # CONDICIONAL POR SI ELIGEN ALGO FUERA DEL MENÚ
    else:
        print("❌ Código de menú incorrecto. Intenta de nuevo.")

    print("\n--------------------------------------------------")

print("\n==================================================")
print("  Estetoscopio fuera de línea. ¡Buen trabajo, Doc! ")
print("    Programa realizado por Arath Valenzuela NL57   ")
print("====================================================")
```
